```python
import os
from agents import Agent, Runner, trace, set_default_openai_client, set_tracing_disabled, set_tracing_export_api_key
from openai import AsyncOpenAI
```

```python
os.environ.get("OPENAI_API_KEY")
```

```python
set_tracing_export_api_key(os.environ.get("OPENAI_API_KEY"))
set_tracing_disabled(False)
```

```python
custom_client = AsyncOpenAI(
    api_key=os.environ.get("GROQ_API_KEY"),  # Your Groq key
    base_url="https://api.groq.com/openai/v1"
)
```

```python
set_default_openai_client(custom_client)
```

```python
agent = Agent(
    name="Jokester", 
    instructions="You are a creative assistant that tells jokes and describes superhero battles", 
    model="llama-3.3-70b-versatile"
)
```

```python
print("\n" + "="*50)
print("COLLECTING 9 JOKES")
print("="*50)
```

```python
joke_messages = []
for i in range(1,10):
    with trace(f"Joke {i}"):  
        result = await Runner.run(agent, "Tell a joke about Autonomous AI Agents")
        joke_messages.append(result.final_output)
        print(f"  ✓ Joke {i} collected")
```

```python
# Print jokes nicely
print("\n" + "="*50)
print("JOKES OUTPUT")
print("="*50)
for i, joke in enumerate(joke_messages, 1):
    print(f"\n--- Joke {i} ---\n{joke}\n")
```

```python
# ===== SUPERHERO FIGHTS COLLECTION =====
print("\n" + "="*50)
print("COLLECTING 9 SUPERHERO FIGHTS")
print("="*50)
```

```python
fight_messages = []
for i in range(1,10):
    with trace(f"Superhero Fight {i}"):  
        result = await Runner.run(agent, "Take 3 characters from DC and Marvel, make them fight, and give a conclusion. All under 100 words per fight.")
        fight_messages.append(result.final_output)
        print(f"  ✓ Fight {i} collected")
```

```python
# Print fights correctly
print("\n" + "="*50)
print("SUPERHERO FIGHTS OUTPUT")
print("="*50)
for i, fight in enumerate(fight_messages, 1):
    print(f"\n--- Fight {i} ---\n{fight}\n")
```

```python
# ===== SUMMARY =====
print("="*50)
print(f"SUMMARY:")
print(f"  • Total jokes collected: {len(joke_messages)}")
print(f"  • Total fights collected: {len(fight_messages)}")
print(f"  • Check traces at: https://platform.openai.com/traces")
print("="*50)
```
