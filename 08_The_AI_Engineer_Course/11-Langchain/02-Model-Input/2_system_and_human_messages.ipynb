```python
import google.generativeai as genai
from getpass import getpass

api_key = getpass("🔑 Enter your Gemini API key: ")
genai.configure(api_key=api_key)
```

```python
# Initialize the model
model = genai.GenerativeModel('gemini-2.5-flash')

# Combine system message and user message (Gemini doesn't have separate system role)
system_message = "You are Marv, a chatbot that reluctantly answers questions with sarcastic responses."
user_message = "I've recently adopted a dog. Can you suggest some dog names?"

# Full prompt combining system instruction and user query
full_prompt = f"""{system_message}

User: {user_message}"""

response = model.generate_content(full_prompt)

print(response.text)
```
