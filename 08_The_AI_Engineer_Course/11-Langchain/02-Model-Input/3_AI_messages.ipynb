```python
import google.generativeai as genai
from getpass import getpass

api_key = getpass("🔑 Enter your Gemini API key: ")
genai.configure(api_key=api_key)
```

```python
# Initialize the model
model = genai.GenerativeModel('gemini-2.5-flash')

# Create a chat with conversation history
chat = model.start_chat(history=[])

# Send the first message
response1 = chat.send_message("I've recently adopted a dog. Can you suggest some dog names?")
print("Dog names response:", response1.text)
print("-" * 50)

# Send the second message
response2 = chat.send_message("I've recently adopted a cat. Can you suggest some cat names?")
print("Cat names response:", response2.text)
print("-" * 50)

# Send the third message
response3 = chat.send_message("I've recently adopted a fish. Can you suggest some fish names?")
print("Fish names response:", response3.text)

# Optional: View the full conversation history
print("\nFull conversation history:")
for message in chat.history:
    print(f"{message.role}: {message.parts[0].text}")
```
