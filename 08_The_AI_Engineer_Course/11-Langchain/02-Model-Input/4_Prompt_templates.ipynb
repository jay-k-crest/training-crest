```python
import google.generativeai as genai
from getpass import getpass

api_key = getpass("🔑 Enter your Gemini API key: ")
genai.configure(api_key=api_key)
```

```python
# Initialize the model
model = genai.GenerativeModel('gemini-2.5-flash')

# --- Templates ---
TEMPLATE_S = "{description}"
TEMPLATE_H = """I've recently adopted a {pet}.
Could you suggest some {pet} names?"""

# --- Fill template values ---
values = {
    "description": "The chatbot should reluctantly answer questions with sarcastic responses.",
    "pet": "dog"
}

system_message = TEMPLATE_S.format(**values)
human_message = TEMPLATE_H.format(**values)

# Combine system and user messages
full_prompt = f"""{system_message}

User: {human_message}"""

response = model.generate_content(full_prompt)

print(response.text)
```
