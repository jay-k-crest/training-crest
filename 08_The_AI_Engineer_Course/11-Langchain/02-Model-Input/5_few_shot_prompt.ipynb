```python
import google.generativeai as genai
from getpass import getpass

api_key = getpass("🔑 Enter your Gemini API key: ")
genai.configure(api_key=api_key)
```

```python
# Initialize the model
model = genai.GenerativeModel('gemini-2.5-flash')

# --- Few-shot examples ---
examples = [
    {
        "pet": "dog",
        "response": 'Oh, absolutely. Because nothing screams "I\'m a responsible pet owner" like asking a chatbot to name your new furball. How about "Bark Twain" (if it\'s a literary hound)?'
    },
    {
        "pet": "cat",
        "response": 'Oh, absolutely. Because nothing screams "I\'m a unique and creative individual" like asking a chatbot to name your cat. How about "Furry McFurFace", "Sir Meowsalot", or "Catastrophe"?'
    },
    {
        "pet": "fish",
        "response": 'Oh, absolutely. Because nothing screams "I\'m a fun and quirky pet owner" like asking a chatbot to name your fish. How about "Fin Diesel", "Gill Gates", or "Bubbles"?'
    }
]

# --- Function to create few-shot prompt ---
def build_few_shot_prompt(pet_name):
    prompt = ""
    
    # Add few-shot examples
    for ex in examples:
        prompt += f"User: I've recently adopted a {ex['pet']}. Could you suggest some {ex['pet']} names?\n"
        prompt += f"Assistant: {ex['response']}\n\n"
    
    # Add the new user query
    prompt += f"User: I've recently adopted a {pet_name}. Could you suggest some {pet_name} names?\n"
    prompt += "Assistant:"
    
    return prompt

# --- Build prompt for a rabbit ---
full_prompt = build_few_shot_prompt("rabbit")

# --- Send to Gemini ---
response = model.generate_content(full_prompt)

# --- Display response ---
print(response.text)
```
