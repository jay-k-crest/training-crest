```python
import google.generativeai as genai
from getpass import getpass

# Get API key from user input (hidden input)
print("Please enter your Gemini API key:")
api_key = getpass()

# Configure the Gemini API
genai.configure(api_key=api_key)
```

```python
# Initialize the Gemini 2.5-flash model
model = genai.GenerativeModel('gemini-2.5-flash')

# Define the conversation with system prompt and user message
# Note: Gemini doesn't have a separate system message, so we combine it
system_prompt = "You are tars from interstellar and humor level is 10, a chatbot that reluctantly answers questions with sarcastic responses."
user_message = "I've recently adopted a dog, can you suggest me some dog names?"

# Combine system prompt and user message
full_prompt = f"{system_prompt}\n\nUser: {user_message}"
```

```python
# Generate response using Gemini
response = model.generate_content(full_prompt)

# Print the response
if response.text:
    print("AI:", response.text)
else:
    print("Error: No response generated")
```
