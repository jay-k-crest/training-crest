```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import re
import pandas as pd
```

```python
data = pd.read_csv("tripadvisor_hotel_reviews.csv")
```

```python
data.info()
```

```python
data.head()
```

```python
data['Review'][0]
```

```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
data.head()
```

```python
en_stopwords = stopwords.words('english')
en_stopwords.remove("not")
```

```python
data['review_no_stopwords'] = [
    ' '.join([word for word in review.split() if word not in en_stopwords])
    for review in data['review_lowercase']
]
```

```python
data.head()
```

```python
data['review_no_stopwords'][0]
```

```python
data['review_no_stopwords_no_punct'] = data.apply(
    lambda x: re.sub(r"[*]", "star", x['review_no_stopwords']),
    axis=1
)
```

```python
data.head()
```

```python
data['review_no_stopwords_no_punct'] = [
    re.sub(r"([^\w\s])", "", text)
    for text in data['review_no_stopwords_no_punct']
]
```

```python
data.head()
```

```python
data['tokenized'] = [
    word_tokenize(text)
    for text in data['review_no_stopwords_no_punct']
]
```

```python
data.head()
```

```python
data['tokenized'][0]
```

```python
ps = PorterStemmer()
```

```python
data["stemmed"] = [
    [ps.stem(token) for token in tokens]
    for tokens in data["tokenized"]
]
```

```python
data.head()
```

```python
data['stemmed'][0]
```

### Lemmatization

```python

lemmatizer = WordNetLemmatizer()
```

```python
data["lemmatized"] = data["tokenized"] \
                    .apply(lambda tokens: \
                    [lemmatizer.lemmatize(token) \
                     for token in tokens])
```

```python
data['lemmatized'][0]
```

```python

data.head()
```

### Ngrams


```python

tokens_clean = sum(data['lemmatized'], [])
```

```python
# unigrams: n=1
unigrams = (pd.Series \
            (nltk.ngrams(tokens_clean, 1)) \
            .value_counts()) 
print(unigrams)
```

```python
# bigrams: n=2
bigrams = (pd.Series \
           (nltk.ngrams(tokens_clean, 2)) \
           .value_counts()) 
print(bigrams)
```

```python
ngrams_4 = (pd.Series \
            (nltk.ngrams(tokens_clean, 4)) \
            .value_counts()) 
print(ngrams_4)
```
